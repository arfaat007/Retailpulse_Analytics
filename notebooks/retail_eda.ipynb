```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ","_").str.replace("-","_")
df.columns
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_month_name'] = df['order_date'].dt.month_name()
df['order_quater'] = df['order_date'].dt.quarter
```

```python
#create business columns

df['profit_margin'] = np.where(
    df['sales'] > 0 ,
    (df["profit"] / df['sales'])*100,
    0
)
```

```python
df['sales_category'] = np.where(
    df['sales'] >= df['sales'].quantile(0.75),
    "high_value",
    np.where(df['sales'] >= df['sales'].quantile(0.40),"medium_value","low_value")
)
```

```python
df['profit_status'] = np.where(df["profit"] >= 0,"Profit","Loss")
```

```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df.to_csv(r"C:\Users\arfat\Downloads\retailpulse-analytics\data\cleaned\retail_cleaned.csv",index=False)
```

```python
#adding kpi's 
total_sales = df['sales'].sum()
total_profit = df['profit'].sum()
total_orders = df['order_id'].nunique()
total_customers = df['customer_id'].nunique()
total_quantity = df['quantity'].sum()
average_order_value = total_sales / total_orders
overall_profit_margin = (total_profit/total_sales)*100

```

```python
print("Total sales: ",round(total_sales,2))
print("total profit: " ,round(total_profit,2))
print("Total orders: " , total_orders)
print("Total  customers: ", total_customers)
print("Total quantity :", total_quantity)
print("Average order Value :", round(average_order_value,2))
print("Overall profit margin :", round(overall_profit_margin,2),"%")
```

```python
#chart 1 monthly sales trend
monthly_sales =( df.groupby(['order_year','order_month'])['sales'].sum().reset_index()
               )

monthly_sales['year_month'] = (
    monthly_sales['order_year'].astype(str)+"-"+
    monthly_sales['order_month'].astype(str).str.zfill(2)
)

plt.figure(figsize = (12,5))
plt.plot(monthly_sales['year_month'],monthly_sales['sales'],marker = 'o')
plt.title("Monthly Sales Trend")
plt.xlabel("month")
plt.ylabel('year')
plt.xticks(rotation= 55)
plt.tight_layout()
plt.savefig(r"C:\Users\arfat\Downloads\retailpulse-analytics\images\charts\montly_sales_trend.png")
plt.show()
```

```python
#chart 2 Sales by region
region_sales = df.groupby("region")['sales'].sum().sort_values(ascending = False)

plt.figure(figsize = (12,5))
region_sales.plot(kind="bar")
plt.title("sales by region")
plt.xlabel("Region")
plt.ylabel("sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(r"C:\Users\arfat\Downloads\retailpulse-analytics\images\charts\sales_by_region.png")
plt.show()
```

```python
#chart 3 profit by category
category_profit = df.groupby('category')['profit'].sum().sort_values(ascending=False)

plt.figure(figsize = (12,5))
category_profit.plot(kind="bar")
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation = 0 )
plt.savefig(r"C:\Users\arfat\Downloads\retailpulse-analytics\images\charts\Profit_by_category.png")
plt.show()
```

```python
#chart 4 sales by customer segment
segment_sales =  df.groupby("segment")['sales'].sum().sort_values(ascending = False)

plt.figure(figsize = (12,5))
segment_sales.plot(kind="bar")
plt.title("sales by customer segment")
plt.xlabel("segment")
plt.ylabel("sales")
plt.xticks(rotation = 0)
plt.savefig(r"C:\Users\arfat\Downloads\retailpulse-analytics\images\charts\sales_by_customer_segment.png")
plt.show()
```
